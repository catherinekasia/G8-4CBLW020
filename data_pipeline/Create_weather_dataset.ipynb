# Create dataset with weather data

This Jupyter notebook loads weather data for all police forces and matches them with data obtained from MET Office weather stations using manually created policeForces-weatherMatching.csv file. The final result is saved in data folder as weather_dataset.csv

```python
import pandas as pd
```

```python
matching = pd.read_csv("../data/policeForces-weatherMatching.csv")
```

```python
matching
```

```python
years = range(2010, 2026)
months = range(1, 13)
columns = ["tmax", "tmin", "af", "rain"]
```

```python
import os

stations = os.listdir("../data/WeatherStations")
stations = [(x.split("."))[0] for x in stations]
stations.sort()

dict = {}

for y in years:
    for m in months:
        for c in columns:
            dict[str(y)+"_"+str(m)+"_"+c] = []

for tmp in ["2026_1_", "2026_2_", "2026_3_", "2026_4_"]:
    for c in columns:
        dict[tmp + c] = []
```

```python
for station in stations:
    print(station)
    f = open("../data/WeatherStations/" + station + ".txt", "r")
    data = [float(x) for x in f.read().split()]
    f.close()

    ind = 2
    for y in range(2010, 2027):
        for m in range(1, 13):
            if y == 2026 and m > 4:
                break
            
            for c in columns:    
                dict[str(y)+"_"+str(m)+"_"+c].append(data[ind])
                ind += 1

            ind += 3
            
```

```python
df = pd.DataFrame(dict, index=stations)
```

```python
df
```

```python
df_copy = df.reset_index()
```

```python
df_copy.rename(columns={'index': 'weatherStation'}, inplace=True)
df_copy
```

```python
result = pd.merge(matching, df_copy, on="weatherStation")
```

```python
result
```

```python
result = result.set_index("policeForce")
result
```

```python
result.to_csv("../data/weather_dataset.csv", index=True)
```
