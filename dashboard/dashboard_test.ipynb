```python
#Dependencies

from ipyleaflet import (Map, basemaps, GeoJSON)

import geopandas as gpd
import json
```

```python
with open("Police_Force_Areas_December_2023_EW_BSC_-3123286944868329346.geojson",'r') as f:
    data = json.load(f)

center = [53,0]
zoom = 5.5
m = Map(basemap=basemaps.CartoDB.Positron, center=center, zoom=zoom)

geo_json = GeoJSON(data=data, 
    style={'color': 'black', 'fillColor': '#E0D071', 'opacity':0.03, 'weight':1.9, 'dashArray':'2', 'fillOpacity':0.6},
    hover_style={'fillColor': '#b08a3e' , 'fillOpacity': 0.8},
    name = 'Countries')
m.add_layer(geo_json)

m
```
